```python
# to make simple calculation
num1 = float(input("Enter first number: "))
num2 = float(input("Entre second number:"))

operation = input("Entre opration(+,-,*,/,%,**,):")

if operation =="+":
    print("result:",num1 + num2)

elif operation =="-":
    print("result:",num1 - num2)

elif operation == "*":
    print("result:",num1 * num2)

elif operation == "/":
    print("result:",num1 / num2)

elif operation == "%":
    print("result:",num1 % num2)

elif operation == "**":
    print("result:",num1 ** num2)

else:
    print("Invalid operation")
```
