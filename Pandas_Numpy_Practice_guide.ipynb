```python
import pandas as pd

data = {
    'Name': ['Amit', 'Sneha', 'Raj', 'Priya'],
    'Marks': [88, 76, 90, 85],
    'Grade': ['B', 'C', 'A', 'B']
}

df = pd.DataFrame(data)

# Replace name
df['Name'] = df['Name'].replace('Amit', 'Aman')

# Add bonus
df['Marks'] = df['Marks'].apply(lambda x: x + 5)

# Grade to remarks
df['Remarks'] = df['Grade'].map({'A': 'Excellent', 'B': 'Good', 'C': 'Average'})

# Result based on updated Marks
df['Result'] = df['Marks'].apply(
    lambda x: 'distinction' if x >= 90 else ('passed' if x >= 80 else 'needs improvement')
)

print(df)
```

```python
import pandas as pd

data = {
    'Department': ['Sales', 'Sales', 'HR', 'HR', 'IT', 'IT', 'Sales', 'HR', 'IT'],
    'Employee': ['Amit', 'Sneha', 'Raj', 'Priya', 'Neha', 'Vikas', 'Karan', 'Divya', 'Meena'],
    'Salary': [50000, 52000, 45000, 47000, 60000, 61000, 53000, 48000, 59000],
    'Age': [25, 27, 24, 26, 28, 30, 29, 23, 31]
}

df = pd.DataFrame(data)
```

```python
salary_dept = df.groupby('Department')['Salary'].sum().sort_values(ascending = False)
print(salary_dept)
```

```python
df.groupby('Department').agg({
    'Age':'mean',
    'Salary':'mean'
})
```

```python
high_salary = df.loc[df['Salary'].idxmax()]
print(high_salary)
```

```python
def Salary_Level(s):
    if s >= 58000:
        return 'high'
    elif s >= 50000:
        return 'medium'
    else:
        return 'low'

df['Salary_Level'] = df['Salary'].apply(Salary_Level)
print(df[['Employee', 'Salary', 'Salary_Level']])
```

```python
count = df['Salary_Level'].value_counts()
print(count)
```

```python
import pandas as pd

# Employee Info
df1 = pd.DataFrame({
    'Emp_ID': [101, 102, 103],
    'Name': ['Amit', 'Sneha', 'Raj'],
    'Department': ['HR', 'Sales', 'IT']
})

# Employee Salary Info
df2 = pd.DataFrame({
    'Emp_ID': [101, 102, 104],
    'Salary': [50000, 52000, 55000]
})

print(df1)
print(df2)
```

```python
result =  pd.concat([df1,df2])
print(result)
```

```python
result1 = pd.concat([df1, df2], axis=1)
print(result1)
```

```python
merged_inner = pd.merge(df1,df2 ,on = 'Emp_ID',how = 'inner')
print(merged_inner)
```

```python
merged_left = pd.merge(df1,df2 ,on = 'Emp_ID',how = 'left')
print(merged_left)
```

```python
merged_right= pd.merge(df1 ,df2,on='Emp_ID',how = 'right')
print(merged_right)
```

```python
merged_outer = pd.merge(df1,df2,on='Emp_ID',how='outer')
print(merged_outer)
```

```python
df1_indexed = df1.set_index('Emp_ID')
df2_indexed = df2.set_index('Emp_ID')

joined = df1_indexed.join(df2_indexed,how='inner')
print(joined)

```

```python
#summarise data like pivot table
import pandas as pd
data = {
    'Product': ['A', 'A', 'B', 'B', 'A'],
    'Region': ['East', 'West', 'East', 'West', 'East'],
    'Sales': [100, 150, 200, 130, 180]
}

df = pd.DataFrame(data)
```
