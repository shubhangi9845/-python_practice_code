```python
# Biggest Number 
def largest_number():
    num1 = int(input())
    num2 = int(input())
    num3 = int(input())

    largest = max(num1,num2,num3)
    print("largest number is",largest)



largest_number()
```

```python
#smallest number
def smallest_number():
    numbers = list(map(int,input("Enter the 5 number here").split()))

    smallest = min(numbers)
    print("smallest number",smallest)

smallest_number()
```

```python
def find_length():
    numbers = list(map(int,input("entre the numbers here with space").split()))
    number_length = len( numbers)
    print("find the length of list",number_length)

find_length()
```

```python
def find_average():
    numbers = list(map(int,input("entre the 5 number here with space").split()))
    find_average =sum(numbers)/len(numbers)
    print("find the average",find_average)


find_average()
```
